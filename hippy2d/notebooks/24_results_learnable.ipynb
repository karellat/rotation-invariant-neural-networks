```python
import seaborn as sns
import pandas as pd
import matplotlib.pyplot as plt
```

```python
file = "wandb_export_2025-10-14T09_21_49.754-06_00.csv"
csv_path = f"/Users/karella/Projects/rotation-invariant-neural-networks/hippy2d/data/exports/{file}"
df = pd.read_csv(csv_path)
df['Name'] = df['Name'].replace({"learnable-smaller-one" : "1", 
                    "learnable-smaller-power" : "$r^n$",
                    "learnable-smaller" : "$r$",
                    "learnable-smaller-square" : "$r^2$",})
df['Preserve'] = df["m_param.layer_kwargs.preserve_energy"] == True
#df = df.loc[df['Preserve'] == True]
```

```python
fig, ax = plt.subplots(figsize=(6,4))
sns.scatterplot(data=df, x="Name", y="test_acc", hue="Preserve", style="Preserve", s=100, ax=ax)
```

```python
fig, ax = plt.subplots(figsize=(6,4))
sns.stripplot(
    data=df,
    x="Name", y="test_acc",
    hue="Preserve", dodge=True, jitter=True, ax=ax
)
```

```python
fig, ax = plt.subplots(figsize=(6,4))
sns.swarmplot(
    data=df,
    x="Name", y="test_acc",
    hue="Preserve", dodge=True, ax=ax
)
```
